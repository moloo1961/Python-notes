# **Blueberry Yield**

## Problem Definition

    The most challenging task in the agricultural sector is to accurately predict crop yield. A typical machine learning algorithm often uses real data to predict crop yield. In this study, I used data generated from a deep learning model trained on the Wild Blueberry Pollination Model, a spatially explicit simulation model validated by field observation and experimental data collected in Maine USA during the last 30 years. The main aim of this study is to leverage the techniques I've learned in regression models to predict blueberry yield. 

    Blueberries are small, round berries with a sweet and slightly tart flavor, belonging to the genus Vaccinium. Native to North America, they have been cultivated for over a century, primarily in two types: highbush (cultivated) and lowbush (wild). Native Americans initially harvested them for food and medicinal purposes, with commercial cultivation beginning in the early 1900s through the efforts of Elizabeth White and Frederick Coville. 

    Today, blueberries are grown worldwide, including emerging farms in Kenya, where the climate and soil conditions are favorable. Blueberries are recognized as a superfood due to their high nutrient and antioxidant content, particularly anthocyanins, and are consumed in various forms, contributing to wellness and disease prevention diets.

## Exploratory Data Analysis

```python
TARGET = "yield"
FEATURES = [c for c in traindata.columns if c not in ["id", TARGET]]
```

```python
corr = traindata.corr(numeric_only=True)[TARGET].sort_values(ascending=False)
print("\nCorrelation with yield:\n", corr)
```

## Feature Engineering and Selection

```python
X = traindata[FEATURES]
y = traindata[TARGET]
X_test_final = testdata[FEATURES]
test_ids = testdata["id"]
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

results = {}
```

```python
def evaluate(name, model, X_val_, y_val_):
    preds = model.predict(X_val_)
    mae = mean_absolute_error(y_val_, preds)

    results[name] = {"MAE": mae}
    print(f"{name:20s} MAE={mae:8.2f}")
```

```python
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_val_s = scaler.transform(X_val)
```

## Model Selection

```python
lr = LinearRegression()
lr.fit(X_train_s, y_train)
evaluate("LinearRegression", lr, X_val_s, y_val)
```

```python
rf = RandomForestRegressor(
    n_estimators=300, max_depth=12, random_state=RANDOM_STATE, n_jobs=-1
)
rf.fit(X_train, y_train)
evaluate("RandomForest", rf, X_val, y_val)
```

```python
xgb_model = xgb.XGBRegressor(
    n_estimators=500, max_depth=4, learning_rate=0.03,
    subsample=0.8, colsample_bytree=0.8, random_state=RANDOM_STATE, n_jobs=-1
)
xgb_model.fit(X_train, y_train)
evaluate("XGBoost", xgb_model, X_val, y_val)
```
