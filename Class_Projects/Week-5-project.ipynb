WHAT IS A CARBON FOOTPRINT?

- A carbon footprint is the total amount of greenhouse gases, especially carbon dioxide, released into the atmosphere due to human activities. 

- Emissions:
    - Transportation (pertol/diesel cars, planes, ships that burn fossil fuels)
    - Electricity Use
    - Food production Impact (farming animals - cows release methane{potent greenhouse gas}, packaging food)
    - Industrial
    - etc

- Measured in metric tons of carbon dioxide.

- Greenhouse gases trap hear in the atmosphere leading to global warming causing problems such as: 
        - hotter summers, 
        - unpredictable weather, melting ice caps that lead to rising sea levels and flooding, natural disasters like hurricanes, drought and wildfires, 
        - animals losing habitats due to climate change and going extinct. 

- Its like a mark we live on the environment, so the bigger our carbon footprint, the more we contribute to climate change.
