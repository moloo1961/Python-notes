WHAT IS A CARBON FOOTPRINT?

- A carbon footprint is the total amount of greenhouse gases, especially carbon dioxide, released into the atmosphere due to human activities. 

- Emissions:
    - Transportation (pertol/diesel cars, planes, ships that burn fossil fuels)
    - Electricity Use
    - Food production Impact (farming animals - cows release methane{potent greenhouse gas}, packaging food)
    - Industrial
    - etc

- Measured in metric tons of carbon dioxide.

- Greenhouse gases trap heat in the atmosphere leading to global warming causing problems such as: 
        - hotter summers, 
        - unpredictable weather, melting ice caps that lead to rising sea levels and flooding, natural disasters like hurricanes, drought and wildfires, 
        - animals losing habitats due to climate change and going extinct. 

- Its like a mark we live on the environment, so the bigger our carbon footprint, the more we contribute to climate change.

## **Food Carbon FootPrint:** *The Environmental Impact of Our Eating Habits*

[nu3](https://www.nu3.de/blogs/nutrition/food-carbon-footprint-index-2018) : Conducted a research on directly comparing different diets in terms of carbon dioxide emissions. The research, which reveals the annual CO2 emissions per person for 130 nations worldwide, shows which countries could significantly reduce their carbon footprint by switching to a plant-based diet, as well as which food types generate the highest carbon dioxide emissions.






```python
# Load libararies

import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import geopandas as gpd
```

```python
# Load data

df = pd.read_csv('food_consumption.csv')
```

### Use the data to explore:

#### 1. East Africa CO2 By Food Category

The average C02 emission per person by food category among five countries (Kenya, Uganda, Tanzania, Rwanda and Ethiopia) located in East Africa.

```python
# Filter East African Countries

e_a_c = ["Kenya", "Uganda", "Tanzania", "Rwanda", "Ethiopia"]

df_e_a = df[df["country"]. isin(e_a_c)]
```

```python
df_e_a.head()
```

```python
# Calculate the average

df_avg_food = df_e_a.groupby("food_category")["co2_emmission"].median().reset_index()
df_avg_food = df_avg_food.sort_values(by="co2_emmission", ascending=False)
print(df_avg_food)
```

```python
# NOTE
# Bar charts handle categorical data (text labels) - cleanly separate a compare distinct items
# FacetGrid handle small multiples
# Line charts handle trends over a continuous period of time
# Pie charts handle data showing how parts add up to a single 100%
# Scatter plots handle relationships or correlations between 2 continous numbers
```

```python
animal_products = ["Beef", "Milk - inc. cheese", "Lamb & Goat", "Fish", "Eggs", "Poultry", "Pork"]
# plant_products = ["Rice", "Wheat and Wheat Products", "Nuts inc. Peanut Butter", "Soybeans"]

color_dict = {
    food: "#A63A50" if food in animal_products else "#3B7A57"
    for food in df_avg_food["food_category"]
}
```

```python
# To create a bar chart

# Size of chart
plt.figure(figsize=(12, 7)) 

# Specify column names
ax = sns.barplot(
    data=df_avg_food,
    y="food_category",
    x="co2_emmission",
    hue="food_category",
    palette=color_dict,
    legend=False
) 

# Clean up chart frame
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Add titles and labels
plt.title("""Animal based food products account for dis-proportionately large amounts of CO2 Emissions\n(As measured by median CO2 emissions)""",
          fontsize=14, pad=15)
plt.ylabel("Food Category", fontsize=12)
plt.xlabel("CO2 Emissions (kg/person/year)", fontsize=12)


plt.tight_layout()
plt.xlim(0, 175)
plt.show
```

#### 5. Box Plot Distribution


A box plot and interpretation of the CO2 contributions of food levels.

```python
# To create a bar chart for each unique food category to its 
# distribution of emissions

# Size of chart
plt.figure(figsize=(14, 7))

# Specify column names
ax = sns.boxplot(
    data=df_e_a,
    y="food_category",
    x="co2_emmission",
    hue="food_category",
    palette="Set3",
    legend=False
) 

# Label the emitters?


# Clean up chart frame
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Add titles and labels
plt.title("Distribution and Variations of CO2 Contributions by Food Levels in East Africa\nHoly cow!",
          fontsize=14, pad=15, weight="bold")
plt.ylabel("Food Category", fontsize=12)
plt.xlabel("CO2 Emissions (kg/person/year)", fontsize=12)


plt.tight_layout()
plt.show
```

#### 6. Top 10 Countries


Top 10 Countries with highest emissions.

```python
df_global = df.groupby("country")["co2_emmission"].sum().reset_index()

df_top10 = df_global.sort_values(by="co2_emmission", ascending=False).head(10)

print(df_top10)
```

```python
# To create a bar chart

# Size of chart
plt.figure(figsize=(12, 6))

# Specify column names
sns.barplot(
    data=df_top10,
    y="country",
    x="co2_emmission",
    hue="country",
    palette="rocket",
    legend=False
) 

# Clean up chart frame
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Add titles and labels
plt.title("Top 10 Countries Globally with Highest Food Carbon FootPrint",
          fontsize=14, pad=15, weight="bold")
plt.ylabel("Country", fontsize=12)
plt.xlabel("Total CO2 Emissions (kg/person/year)", fontsize=12)


plt.tight_layout()
plt.show
```
