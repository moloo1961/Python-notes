# Number Guessing Game Application

Build a Guess the Number (with Difficulty Levels) Game

Create a guessing game where the computer generates a random number and the user tries to guess it

```python
# Generate secret number
import random

chosen_answer = random.randint(1,100)
print(chosen_answer) # make it a variable 

```

```python
print("Welcome to the Number Guessing Game\n")
Difficulty_levels = ("A. Easy\nB. Medium\nC. Hard\n")

print(f"Choose Difficulty Level:\n{Difficulty_levels}") 

A = "A"
B = "B"
C = "C"

# Setup difficulty levels
while True: 
    level = input("Choose the difficulty level:").upper().strip()
    try:
        if level not in ["A", "B", "C"]: # 
                raise ValueError("INVALID! CHOOSE A, B, or C.")
        if level == A:
            attempts = 10
            print("You have chosen Easy!")
            print(f"You have {attempts} attempts to guess a number between 1 and 100")
    
        elif level == B:
           attempts = 7
           print("You have chosen Medium!")
           print(f"You have {attempts} attempts to guess a number between 1 and 100")
        else: # Runs if level is C
           attempts = 5
           print("You have chosen Hard!")
           print(f"You have {attempts} attempts to guess a number between 1 and 100")

        break # Stops loop once valid level is picked
    except ValueError as e: # If error happens
            print(e)
```

```python
while attempts > 0: # Loop that controls the game
    answer_display = int(input("This is my number")) # Guess in loop
    # Test answer immediately at start of loop
    if answer_display == chosen_answer:
        print("Hooray!! You guessed right!")
        break # Stop loop incase answer is correct
    elif answer_display > chosen_answer:
        print("Too High! Keep Guessing")
    else:
        print("Too Low! Keep Guessing")
    attempts -= 1 # Reduce attempts after wrong guess
    print(f"> Attempts left: {attempts}\n")
    # Run if loop ends naturally i.e attempts == 0
else:
    print(f"Game Over! The number was: {chosen_answer}") 
    print("You Lost")
```
