## **No SQL**

Not Only SQL

It's a type of database designed for:
- Flexible data.
- Fast scaling.
- Huge amounts of information.
- Modern websites/apps.

Unlike SQL databases, NoSQL databases do not always use tables and rows.

SQL vs NoSQL

SQL -                                   NoSQL
- Tables                               - Flexible structures

- Rows and columns                     - Documents or key-values or graphs

- Fixed schema                         - Dynamic schema

- Structured data                      - Unstructured data

- SQLite, MySQL                        - MongoDB, Firebase 

- Best for banking                     - Best for apps or social media

- Scalabilty(vertical-big server)      - (Horizontal- more servers)

REAL-WORLD EXAMPLES

SQL

- Best for: hospitals, banks, school systems, payroll.

- This is because relationships matter.


NoSQL

- Best for: chat apps, social media, analytics, AI apps, recommendation systems.

- This is because data changes constantly.

MAIN TYPES OF NoSQL

1. Document Databases

- Stores data like JSON documents.
- This is powerful because different users can have different fields.
- Also no need for identical columns.
- Example: MongoDB

Instead of tables, example:

2. Key-Value Databases

- It's very fast.
- Used for: caching, sessions, settings.
- Example: Redis

3. Graph Databases

- Stores relations.
- Used for: social networks, recommendations.
- Example: Neo4j

4. Column Databases

- Stores data by columns instead of rows.
- Used for: analytics, huge datasets.
- Example: Cassandra

```python
import os
from dotenv import load_dotenv
from pymongo import MongoClient

# Secret Password
load_dotenv()
uri = os.getenv("MONGO_URI") # Connection str
client = MongoClient(uri) # Creates actual connection to MongoDB

db = client["school"]
students =db["students"]

try:
    client.admin.command('ping')
    print("Successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

```python
# Insert one

student_001 = {
    "name": "Takopi San",
    "age": 16,
    "gpa": 3.8,
    "courses": ["Math", "Physics"],   # Lists are allowed!
    "address": {"city": "Osaka"}   # Nested Dicts are allowed!
}

result = students.insert_one(student_001)
print(f"Created Student ID: {result.inserted_id}")
```

```python
# Insert many

batch_data = [
    {"name": "Megumi", "age": 18, "gpa": 4.0},
    {"name": "Sukuna", "age": 17, "gpa": 2.5},
    {"name": "Jogo", "age": 17, "gpa": 2.4},
    {"name": "Mahito", "age": 19, "gpa": 1.9}
]

result = students.insert_many(batch_data)
print(f"Created {len(result.inserted_ids)} students.")
```

```python
# Read (Query Operations)

# .find()

# Find Sukuna

sukuna = students.find_one({"name": "Sukuna"})
print(sukuna)
```

```python
# Find Many (Get List)
# Find all students with GPA > 3.0
# Syntax: { field: { $operator: value } }
high_achievers = students.find({"gpa": {"$gt": 3.0}})

print("High Achievers:")
for student in high_achievers:
    print(f"- {student['name']} (GPA: {student['gpa']})")
```

- Common Query Operators
    - $eq : Equal to
    - $gt : Greater than
    - $lt : Less than
    - $gte : Greater than or Equal to
    - $lte : Less than or Equal to
    - $in : Match any value in a list ({"age": {"$in": [14, 15]}})
    - $set : Only change certain field

```python
# Update

# Simple Update ($set)

# Change Takopi's age to 17
students.update_one(
    {"name": "Takopi San"},       # Filter (Who?)
    {"$set": {"age": 17}}       # Action (What?)
)
```

```python
# List Modification ($push)

students.update_one(
    {"name": "Takopi San"},
    {"$push": {"courses": "Chemistry"}} # $push appends values to lists
)
```

```python
# Increment ($inc)

# It's Takopi's birthday! Add 1 to his age.
students.update_one(
    {"name": "Takopi San"},
    {"$inc": {"age": 1}} # $inc : selects and updates values
)
```

```python
# Delete 

# Delete one
students.delete_one({"name": "Jogo"})
```

```python
# Delete many

# Delete everyone with a low GPA
# Everyone with gpa < 2.0
result = students.delete_many({"gpa": {"$lt": 2.0}})
print(f"Expelled {result.deleted_count} student.")
```

```python
# A bookstore wants to store book records, including title, 
# author, and price, in MongoDB Atlas.

# 1. Set Up MongoDB Collection

# Create a database named bookstore.
# Create a collection named books.

# 2. Insert Sample Data

# Insert at least 5 book records with the following fields:
# title
# author
# price

# 3. Perform CRUD Operations

# Read: Fetch all books where the price is greater than 10.
# Update: Change the price of a specific book.
# Delete: Remove a book from the collection by its ID or title.
```

```python
import os
from dotenv import load_dotenv
from pymongo import MongoClient

# Set up: Connect and Create Database/Collection
# Secret Password
load_dotenv()
uri = os.getenv("MONGO_URI")
client = MongoClient(uri)

try:
    client.admin.command('ping')
    print("Successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

```python
# 1. Set Up MongoDB Collection

# Create a database named bookstore.
# Create a collection named books.
db = client["bookstore"]

bookstore =db["books"]
```

```python
# 2. Insert Sample Data

# Insert at least 5 book records with the following fields:
# title
# author
# price

# Insert sample books
sample_books = [
    {"title": "Things Fall Apart", "author": "Chinua Achebe", "price": 15},
    {"title": "Half of a Yellow Sun", "author": "Chimamanda Ngozi Adichie", "price": 10},
    {"title": "The Hobbit", "author": "J.R.R. Tolkien", "price": 20},
    {"title": "Born a Crime", "author": "Trevor Noah", "price": 8},
    {"title": "MongoDB Guide", "author": "Data Guru", "price": 25}
]
bookstore.insert_many(sample_books)
```

```python
# 3. Perform CRUD Operations

# Read: Fetch all books where the price is greater than 10.

expensive_books = bookstore.find({"price": {"$gt": 10}})
for book in expensive_books:
    print(f"-{book['title']}: ${book['price']}")
```

```python
# Update: Change the price of a specific book.
bookstore.update_one(
    {"title": "The Hobbit"},       # Filter (Who?)
    {"$set": {"price": 22}}       # Action (What?)
)
```

```python
# Delete: Remove a book from the collection by its ID or title.
bookstore.delete_one({"title": "Things Fall Apart"})
```

```python
result = bookstore.delete_many({"price": {"$gt": 5}})
print(f"Books removed from catalog: {result.deleted_count}.")
```
