## Advanced Concepts

### Named Groups (The Maintainable Way)

- Allows you to label captured text using (?P< name >pattern) instead of relying on index numbers.

- ?P : tells python this is a special extension
- < name > : is the label you assign
- ... : is the actaual pattern


```python
import re

log_entry = "Event: LOGIN_SUCCESS User: admin_01 Time: 14:30"

# FRAGILE WAY (Numbered Groups)
# If we add anything to the start of this pattern, indexes shift.
fragile_pattern = r"Event: (\w+) User: (\w+) Time: (\d+:\d+)"

# PROFESSIONAL WAY (Named Groups)
# We embed the keys 'event', 'user', and 'time' into the pattern.
robust_pattern = r"Event: (?P<event>\w+) User: (?P<user>\w+) Time: (?P<time>\d+:\d+)"

match = re.search(robust_pattern, log_entry)

if match:
    # 1. Access by Name (Readable)
    print(f"User: {match.group('user')}")
    print(f"Action: {match.group('event')}")
    
    # 2. Convert to Dictionary (Powerful)
    # This creates: {'event': 'LOGIN_SUCCESS', 'user': 'admin_01', 'time': '14:30'}
    data = match.groupdict() 
    print(data['time'])
```

### Flags (Handling Real-World Data)

- By default python's regex is very strict.
- Flags let you relax the rules.

re.IGNORECASE OR re.I

- Tells python to stop caring about capital vs lowercase letters.

- Without it - Searching for "Python" will not find "python"
- With it - Matches both. (Great for user input)

re.DOTALL OR re.S
- Changes how the dot . works

- The problem : dot . matches everything except a \n
- The fix : re.DOTALL forces the . to match new lines too (Great for scraping big blocks of text from websites)

re.MULTILINE OR re.M

- Changes how the "start" ^ and "end" $ symbols work when you have a block of text with many lines

- Without it - Python treats the entire block of text as one single, long line.
- With it - Python treats every line as its own starting and ending point.

```python
# Combining flags
import re

text = """
Start
Title: PYTHON 
Description: Python is a
versatile language.
End
"""
# Scenario: Extract everything between 'Title:' 
# and 'End', regardless of case or newlines.
```

```python
# 1. Case Insensitive (Matches 'PYTHON' even though we wrote 'python')
# 2. Dot All (Matches the newlines inside the description)
pattern = r"title: (.*?) end"

match = re.search(pattern, text, re.IGNORECASE | re.DOTALL)
if match:
    print(match.group(1).strip())
```

### Data Cleaning with re.sub()

- Data cleaning fixes 'messy' text using re.sub() (Basically Substitute)
- Regex does not just find data, it is the industry standard tool for cleaning data.
- Function : re.sub(pattern, replacement, string)
- Logic : 'Find every instance of X and replace it with Y.'

```python
# Basic Substitution - replacing messy delimiters or bad characters
```

```python
text = "Item#1; Item#2; Item#3"

# Replace both '#' and ';' with a simple hyphen

clean = re.sub(r"[#;]", "-", text)
print(clean) 
```

```python
sample = "I have a yellow and blue sweater"

# Blue to purple using regular expression

import re

sample = "I have a yellow and blue sweater"

# Use re.sub i.e pattern, replacement, original_text
new_sample = re.sub(r"blue", "purple", sample)

print(new_sample)
```

## Web crawling 

```python
import requests

url = "https://www.example.com"

# The "Costume" - This tells the server we are a Windows 10 PC using Chrome.
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36 Edg/147.0.0.0'
} 

# Sending the request with our costume on
response = requests.get(url, headers = headers)
```

```python
from bs4 import BeautifulSoup

# Step 1: Feed the HTML content to the Soup
# We use response.content (bytes) instead of text to avoid encoding errors
soup = BeautifulSoup(response.content, "html.parser")

# Now 'soup' is a navigable tree!
```

```python
# Find the first <h1> tag
header = soup.find("h1")

# Find a div with a specific ID
main_content = soup.find("div", id="main-content")

print(header)
print(main_content)
```

```python
# Returns a list: [<a href="...">Link 1</a>, <a href="...">Link 2</a>]
all_links = soup.find_all("a")

# Looping through the list
for link in all_links:
    print(link.text)
```

### Example

```python
import requests
from bs4 import BeautifulSoup

coke_url = "https://www.kenyans.co.ke/news" # Request page

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36 Edg/147.0.0.0'
} # add costume

response = requests.get(coke_url, headers = headers) # Get and read page content

soup = BeautifulSoup(response.content, "html.parser")

titles = soup.find_all("h2") [:4] # Slice to find article headers 

print(titles)
```

```python
for t in titles:
    print(t.text.strip()) # Clean to get text
```
