# CLI Quiz Game Application

Build a Command-Line Based Quiz Application

Create a list of questions with Question text, Multiple choice (A,B,C,D), The correct answer

```python
# Introduce the game
print("Welcome to the Find a Genius Quiz App")
print("Are you ready to begin?")
print("A) Yes")
print("B) No")

answer = input("Select an option (A) Yes or B) No):")

if answer == "A":
    print("Let's begin!")
else:
    print("Goodbye") # Game does not proceed
```

```python
# Questions in a dictionary
# Arrange them as key: value collection
que = "question"
cho = "choices"
ans = "answer" # variables to represnt the values in the dictionary
def add_questions(que, cho, ans): 
    return {"question": que, "choices": cho, "answer": ans} # Save result
Questions = [add_questions("What is the capital of France?", "A) Paris B) Lyon C) Marseille D) Nice", "A"),
             add_questions("H2O is known as?", "A) Hydrogen B) Sulphur C) Oxygen D) Water", "D"),
             add_questions("2 + 2?", "A) 4 B) 3 C) 5 D) 6", "A"),
             add_questions("Author of 1984?", "A) Alldous Huxley B) George Orwel C)Ray Bradbury D) Lana del Rae", "B"),
             add_questions("What is the name of the largest planet?", "A) Earth B) Mars C) Jupiter D) Venus", "C")]
```

```python
# A way to tally score
score = 0
```

```python
import random # Mix order for every new run
random.shuffle(Questions) 

for que in Questions:
    print(f"\n{que['question']}") 
    print(que["choices"]) 

    while True: # Weed out invalid input
        answer = input("Choose an answer (A,B,C or D):").upper().strip()

        try:
            
            if answer not in ["A", "B", "C", "D"]:
                raise ValueError("INVALID! CHOOSE A, B, C or D.") # Dicatate what constitutues an error
            
            
            if answer == que["answer"]:
                print("CORRECT")
                score += 1 # Sum up scores when correct
            
            else:
                print("INCORRECT") 

            break # stop 
        except ValueError as e: # If error happens
            print(e)
```

```python
# To show score change
print(f"\nFinal score: {score}") 
```

```python
# Give feedback
if score == 0:
    print("Try again")
elif score >= 4:
    print("GOOD")
elif score == 5:
    print("EXCELLENT")
else:
    print("YOU CAN DO BETTER") 
```
