## Multi-Line Strings

```python
# We use multi-line strings to create a string variable that spans multiple lines.

mokimo_recipe = """"
1. Boil the potatoes until soft 
2. Mash potatoes and add some butter, salt and pepper to taste
3. Serve the mashed potatoes with your choice of protein and vegetables
"""
```

```python
# You can make a multi-line string into upper case using .upper() or lower case using .lower()

mokimo_recipe = mokimo_recipe.upper()
print(mokimo_recipe)
```

**Key points on Multi-Line Strings**

- Enhances readability
- Used in documentation
- Used for longer text such as instruction or error messages

## Data Structures in Python

### Lists in Python

- Lists store multiple values in a single varaiable
    - Lists can contain any combination of data type e.g 'boolean' , 'string' , 'int'

- Lists are ordered
    - One can use subsetting or indexing
    - Python starts counting from **zero** for the first element

```python
fuel_prices = [90, 150, 100, 194, 60, 76]
print(fuel_prices)
```

### Accessing Elements of a List

- We use the square brackets to subset - [ ]
- Sub-setting syntax : name_of_list[index]

```python
fuel_prices
```

```python
fuel_prices[2]

# Counts from left to right 0,1,2
```

```python
fuel_prices[-4]

# Counts from right to left -1,-2,-3,-4
```

### Slicing in Lists / Accessing Multiple Elements in a List


```python
fuel_prices
```

```python
fuel_prices[2:5]

# To index from 100 - 60(middle elements), put the first index and the index after what you want
# 2 for 100 and 5 for the number that comes after 60
# remember to count index from 0
```

```python
fuel_prices[:3]

# Get the first three fuel prices
# These are values from index zero to index two (not inclusive of index three)
```

```python
fuel_prices[3:]

# Get the the elements from the fourth index onwards
# Always skip the value you actually want
```

#### Alternating Access

```python
fuel_prices
```

```python
# Accessing every other element in the list
# Add a step-size to the slicing
# To go skipping
fuel_prices[::2]
```

```python
# Accessing every other element starting from the second element (index 1)
fuel_prices[1::3]
```

```python
# Without a value at the beginning inside the brackets the index 1 starts from the first number
# not index zero but one i.e 1 is 90 skip, 2 is 100 skip, 3 is 60
fuel_prices[::3]
```

## Tuples in Python

- A tuple is an ordered, built-in collection type used to store multiple items in a single variable
- Tuples are immutable. 
- Once created, contents cannot be changed, added to or removed.

```python
# Explanation with code

# Creating a tuple
num_tuple = (10,20,30)
print(f"Original tuple: {num_tuple}")
```

```python
# An element work like a list

print(f"First element: {num_tuple[0]}")
```

```python
# Attempting to change element
try:
    num_tuple[0] = 100
except TypeError as e:
    print(f"Error when trying to modify: {e}")
```

- You must create a tuple to change it

```python
# Methods used on tuples

num_tuple = (10,20,30)
print(num_tuple.count(2)) # count 
```

## Sets in Python

- Contain unique data
- They are unchangeable
    - You can add or remove but you can't change what's already there

set = { }

```python
mokimo_ingredients = {"Potatoes", "Pumpkin leaves", "Magarine", 
                      "Salt", "Pepper","Peas", "Beans", "Carrots", "Onions"}
print(mokimo_ingredients)

# Sets have curly braces
```

```python
mokimo_ingredients = ["Potatoes", "Pumpkin leaves", "Magarine", 
                      "Salt", "Pepper","Peas", "Beans", "Carrots", "Onions"]
print(mokimo_ingredients)

# lists have square brackets
```

```python
# Changing a list to as set
mokimo_ingredients_list = ["Potatoes", "Pumpkin leaves", "Magarine", 
                           "Salt", "Pepper","Peas", "Beans", "Carrots", "Onions"]
print(mokimo_ingredients_list)
```

```python
mokimo_ingredients_set_from_list = set(mokimo_ingredients_list)
print(mokimo_ingredients_set_from_list)
```

```python
print(type(mokimo_ingredients_set_from_list))
```

```python
# Sorting my set
sorted_mokimo_ingredients = sorted(mokimo_ingredients_set_from_list)
print(sorted_mokimo_ingredients)
```
