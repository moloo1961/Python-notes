```python
import os
from dotenv import load_dotenv
from pymongo import MongoClient

# Secret Password
load_dotenv()
uri = os.getenv("MONGO_URI")

client = MongoClient(uri)

db = client["school"]

students =db["students"]
try:
    client.admin.command('ping')
    print("Successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

```python
# Insert one

student_001 = {
    "name": "Takopi San",
    "age": 16,
    "gpa": 3.8,
    "courses": ["Math", "Physics"],   # Lists are allowed!
    "address": {"city": "Osaka"}   # Nested Dicts are allowed!
}

result = students.insert_one(student_001)
print(f"Created Student ID: {result.inserted_id}")
```

```python
# Insert many

batch_data = [
    {"name": "Megumi", "age": 18, "gpa": 4.0},
    {"name": "Sukuna", "age": 17, "gpa": 2.5},
    {"name": "Jogo", "age": 17, "gpa": 2.4},
    {"name": "Mahito", "age": 19, "gpa": 1.9}
]

result = students.insert_many(batch_data)
print(f"Created {len(result.inserted_ids)} students.")
```

```python
# Read (Query Operations)

# .find()

# Find Sukuna

sukuna = students.find_one({"name": "Sukuna"})
print(sukuna)
```

```python
# Find Many (Get List)
# Find all students with GPA > 3.0
# Syntax: { field: { $operator: value } }
high_achievers = students.find({"gpa": {"$gt": 3.0}})

print("High Achievers:")
for student in high_achievers:
    print(f"- {student['name']} (GPA: {student['gpa']})")

```

- Common Query Operators
    - $eq : Equal to
    - $gt : Greater than
    - $lt : Less than
    - $gte : Greater than or Equal to
    - $lte : Less than or Equal to
    - $in : Match any value in a list ({"age": {"$in": [14, 15]}})

```python
# Update

# Simple Update ($set)

# Change Takopi's age to 17
students.update_one(
    {"name": "Takopi San"},       # Filter (Who?)
    {"$set": {"age": 17}}       # Action (What?)
)
```

```python
# List Modification ($push)

students.update_one(
    {"name": "Takopi San"},
    {"$push": {"courses": "Chemistry"}}
)
```

```python
# Increment ($inc)

# It's Takopi's birthday! Add 1 to his age.
students.update_one(
    {"name": "Takopi San"},
    {"$inc": {"age": 1}}
)
```

```python
# Delete 

# Delete one
students.delete_one({"name": "Jogo"})
```

```python
# Delete many

# Delete everyone with a low GPA
result = students.delete_many({"gpa": {"$lt": 2.0}})
print(f"Expelled {result.deleted_count} students.")
```
