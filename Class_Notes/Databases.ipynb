#  **SQL Fundamentals**                   

```python
import sqlite3 # To access SQL

database =sqlite3.connect("Students.db") # to connect to database
connecting = database.cursor() 
create_a_table = connecting.execute("""
CREATE TABLE Students(
id INTEGER PRIMARY KEY AUTOINCREMENT,
first_name TEXT NOT NULL,
last_name TEXT NOT NULL,
age INTEGER,
grade TEXT,
email TEXT UNIQUE,
gpa REAL DEFAULT 0.00
)
""")
```

```python
searching = connecting.execute("SELECT * FROM Students WHERE age < 25")
print(searching.fetchone())

database.commit()
database.close()

# print(searching.fetchall())

# print(searching.fetchmany(1)) WHERE age = 10
```
