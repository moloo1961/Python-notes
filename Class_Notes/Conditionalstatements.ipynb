# Conditional Statements

Conditional statements are statements that perform a specific task if the condition is True and if it is False then do nothing.

if > condition is met > Perform action

```python
# As a Vendor at a concert you have a target number of units you should sell
sales_target = 300
units_sold = 279

# Let me Compare the sales

if units_sold >= sales_target:
    print("Target Achieved")

# Check if we were close to the target
elif units_sold >= 250:
    print("Target almost Achieved") # If first  condition is achieved, python will stop at that

# Otherwise...
elif units_sold == 290:
    print("10 units left to hit target")

elif units_sold == 279:
    print("279 units sol, 21 units left to hit target")

else:
    print("Target not Achieved")
```

You can use as many elif statements as you like

## Conditional Statements in for Loops

```python
feedback_scores
```

```python
for score in feedback_scores:
    # Check if the score is above 50
    if score > 50:
        print(score)
```

```python
feedback_scores = [10, 45.0, 55.9, 60, 30.56, 91.1, 75.6, 80.6, 12.9]
```

```python
count = 0
```

```python
for score in feedback_scores:
    # Use for loop to count and print scores > 50
    if score > 50:
        print(score)
        count += 1
print(f"Total scores above 50: {count}")
```

```python
for score in feedback_scores:
    # Check if even 
    if score % 2 == 0:
        print(score)
        count += 1
print(f"Total even numbers: {count}")
```

# While Loops


While 'condition':
    action


While is used mostly for continous tasks:
- Accelerate 'while' a button is being pressed
- Monitor 'while' the levels are below or above a specific threshold

```python
# Declare a variable to store the stock total

tech_books_stock = 20

# Declare a variable to store the number of purchases

tech_books_purchases = 0

# 'while' there are still books in stock keep selling

while tech_books_purchases < tech_books_stock:
    # Simulate a purchase by incrementing the number of purchases
    tech_books_purchases +=

    # After each sell print the remaining stock
    print(tech_books_stock - tech_books_purchases)
    
```

### Sample Keywords in python

- 'in' = Checks if a value is in a variable is in a variable or a Data Structure (List, Tuple, Dictionary etc)

- 'not' = Checks if a condition is not met

- 'and' = Checks if multiple conditions are met

- 'or' = Checks if one (**or more**) is met

### The 'in' keyword

```python
# The 'in' Keyword
# Used in this dictionary
# key:value
ArtemissII_Astronauts = {"Reid": 39,
                         "Chari": 42,
                         "Kim": 25}
```

```python
# Print the keys of the 'Artemis_Astronauts' Dictionary
print(ArtemissII_Astronauts.keys())
```

```python
# Print the values of the 'Artemis_Astronauts' Dictionary
print(ArtemissII_Astronauts.values())
```

```python
ArtemissII_Astronauts_Detailed = {"Reid": {"Age": 39, "Role": "Commander"},
                                 "Chari": {"Age":42, "Role": "Pilot"},
                                 "Kim": {"Age": 25, "Role": "Mission Specialist"}}
```

```python
# Print the keys of the ArtemissII_Astronauts_Detailed
print(ArtemissII_Astronauts_Detailed.keys())
```

```python
# Print the values of the ArtemissII_Astronauts_Detailed
print(ArtemissII_Astronauts_Detailed.values())
```

```python
# Check if "Reid" is in 'ArtemisII_Astronauts' Dictionary
if "Reid" in ArtemissII_Astronauts.keys():
    print(True)

else:
    print(False)
```

### The 'not in' keyword

```python
# Check if "Reid" is not in 'ArtemisII_Astronauts' Dictionary
if "Reid" not in ArtemissII_Astronauts.keys():
    print(True)

else:
    print(False)
```

### The 'and' keyword

```python
# Use the 'and' keyword to check if "Kim" is a key in the ArtemisII_Astronauts Dictionary
# and if the minimum age of all the astronauts is > 20

if "Kim" in ArtemissII_Astronauts.keys() and min(ArtemissII_Astronauts.values()) > 20:
    print(True)
else:
    print(False)
```

### The 'or' keyword

```python
# Use the 'or' keyword to check if "Chari" is a key in the ArtemisII_Astronauts Dictionary
# and if the minimum age of all the astronauts is < 20

if "Chari" in ArtemissII_Astronauts.keys() or min(ArtemissII_Astronauts.values()) > 20:
    print(True)
else:
    print(False)
```
