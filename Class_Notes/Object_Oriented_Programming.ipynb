## Encapsulation

- Encapsulation is the practice of bundling Data (Attributes) and Methods (Behaviors) into a single unit (a Class) and restricting direct access to some of that object's components.

```python
# Public Members
# Accessible anywhere

# self.name (No underscore - Open to everyone)
class Employee:
    def __init__(self, name, salary):
        # Public Attribute (No underscores)
        self.name = name 
        self.salary = salary 

    def show_details(self):
        # Accessible inside the class
        print(f"Name: {self.name}, Salary: {self.salary}")

# Usage
emp = Employee("James", 100000)

# Accessible outside the class
print(emp.name) # Output: James
emp.show_details()
```

```python
# Protected Members
# Accessible inside the class and its subclasses

# self._project (one underscore - Internal use only)
# 1. Parent Class
class Company:
    def __init__(self):
        # Protected Member (One Underscore)
        self._project = "AI Development"

# 2. Child Class
class Employee(Company):
    def __init__(self, name):
        self.name = name
        # Industry Standard: Use super() to initialize parent
        super().__init__()

    def show_project(self):
        # Accessing protected member in child class -> PERMITTED
        print(f"{self.name} is working on: {self._project}")

# Usage
c = Employee("James")
c.show_project()

# NOTE: This works, but is considered 'Bad Practice' in the industry:
print(c._project) 
```

```python
# Private Members
# Accessible only inside the class where they are defined

# self.__salary (two underscores - Do not touch)
class Employee:
    def __init__(self, name, salary):
        self.name = name
        # Private Member (Two Underscores)
        self.__salary = salary

    # Public method allows us to VIEW the private data safely
    def get_salary(self):
        print(f"Salary is: {self.__salary}")

# Usage
emp = Employee("James", 100000)

# 1. Accessing Public Data -> Works
print(emp.name) 

# 2. Accessing Private Data -> FAILS
# print(emp.__salary) 
# AttributeError: 'Employee' object has no attribute '__salary'

# 3. Accessing via a Method -> Works
emp.get_salary() # Output: Salary is: 100000
```

The Comprehensive Example
- Combines all three levels to show a realistic class structure

```python
class Employee:
    def __init__(self, name, id_number, salary, project):
        # 1. Public: Everyone can see the name
        self.name = name
        
        # 2. Public: Everyone can see the ID
        self.id = id_number
        
        # 3. Protected: Only subclasses (like Manager/Developer) should touch this
        self._project = project
        
        # 4. Private: NO ONE should see this directly. It is sensitive.
        self.__salary = salary

    # Public method to display details (The "Interface")
    def show_details(self):
        # Inside the class, we can access EVERYTHING (even private)
        print(f"Employee: {self.name} (ID: {self.id})")
        print(f"Project: {self._project}")
        # We can view private data internally
        print(f"Salary: ${self.__salary}")

# --- EXECUTION ---
emp1 = Employee('James', 102, 100000, 'Python Migration')

# 1. Safe Access
emp1.show_details()

# 2. Direct Access Attempts
print(f"Public Name: {emp1.name}")       # Works
print(f"Protected Project: {emp1._project}") # Works (But warns developers)
# print(f"Private Salary: {emp1.__salary}") # CRASH (AttributeError)
```

## Abstraction

Abstraction = hide the complicated stuff, show only what matters

  - You use something without knowing how it works inside

  - You focus on what it does, not how it does it

#### Simple Real Life Example
  
Think of:
An ATM machine -
You insert card → withdraw money

You don’t know:

    - How the bank verifies your PIN
    - How the system connects to servers

    You just use it–-That’s abstraction.

Abstraction in Python means:

> “I will create a rule/template, and force > other classes to follow it.”

- How to Achieve Abstraction in Python

Unlike Java, Python does not have built-in abstract classes by default. We must import the abc module (Abstract Base Classes).

#### The Three Rules of Abstraction:

    1. Inheritance: You must inherit from ABC.
    2. Decorator: You must use @abstractmethod on methods you want to enforce.
    3. No Instantiation: You cannot create an object from an Abstract Class. If you try, Python will throw an error.


```python
# The Automobile Standard
# Car is the abstract concept. You cannot buy a generic car 
# You must buy a specific brand

# 1. Import the necessary tools
from abc import ABC, abstractmethod

# 2. Define the Abstract Base Class
class Car(ABC):
    
    # This acts as a strict rule. 
    # Any subclass MUST implement this method, or Python will crash.
    @abstractmethod
    def mileage(self):
        pass

# 3. Create Concrete Classes (The Implementation)
class Tesla(Car):
    def mileage(self):
        print("The mileage is 30kmph (Electric)")

class Suzuki(Car):
    def mileage(self):
        print("The mileage is 25kmph (Petrol)")

class Duster(Car):
    def mileage(self):
        print("The mileage is 24kmph (Diesel)")

class Renault(Car):
    def mileage(self):
        print("The mileage is 27kmph (Hybrid)")

# --- EXECUTION ---

# 1. Trying to create the abstract class directly will FAIL
# c = Car() 
# Error: Can't instantiate abstract class Car with abstract method mileage

# 2. Creating Concrete Objects
t = Tesla()
s = Suzuki()
d = Duster()
r = Renault()

# 3. Calling the methods
t.mileage()  # Output: The mileage is 30kmph (Electric)
s.mileage()  # Output: The mileage is 25kmph (Petrol)
d.mileage()  # Output: The mileage is 24kmph (Diesel)
r.mileage()  # Output: The mileage is 27kmph (Hybrid)
```

## Inheritance

```python
# Inheritance - Reusing and Extending

# Inheritance lets a child class reuse everything from a parent 
# class and add or overside as needed.

# This enforces the DRY principle (Don't Pepeat Yourself)

# PARENT (BASE) CLASS
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.is_alive = True

    def eat(self):
        print(f"({self.name} is eating)")

    def sleep(self):
        print(f"{self.name} is sleeping")
            
    def __str__(self):
        return f"{self.__class__.__name__}: {self.name}, Age {self.age}"
        
# CHILD CLASS
class Dog(Animal): # Dog inherits everything from Animal
    def __init__(self, name, age, breed):
        super().__init__(name, age) # Call parent's __init__first
        self.breed = breed # Add Dog-specific attribute

    def bark(self): # New method unique to Dog
        print(f"{self.name} says: Woof!")

    def fetch(self, item):
        print(f"{self.name} fetches the {item}!")

class Cat(Animal): 
    def __init__(self, name, age, is_indoor):
        super().__init__(name, age) 
        self.is_indoor = is_indoor

    def purr(self):
        print(f"{self.name} purrs: Contendedly.")

    def fetch(self): # OVERRIDE parent method with new behaviour
        super().sleep() # Still call parent's version first
        print(f" ...{self.name} sleeps for 16 hours straight.")

# MULTI-LEVEL INHERITANCE
class GuideDog(Dog): # GuideDog inherits from Dog which inherits from Animal
    def __init__(self, name, age, breed, owner):
        super().__init__(name, age, breed) 
        self.breed = owner 

    def guide(self):
        print(f"{self.name} carefully guides {self.owner} across the street.")
```
