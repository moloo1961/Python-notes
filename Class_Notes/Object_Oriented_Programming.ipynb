```python
class Employee:
    """
    A class to represent an employee.

    States :
    - name 
    - salary

    Behaviours:
    - file returns
    """
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
```

```python
emp001 = Employee("Isabel Merie", 890000)
emp001.name
```

```python
emp002 = Employee("Itangishaka John", 450000)
emp002.name
```
