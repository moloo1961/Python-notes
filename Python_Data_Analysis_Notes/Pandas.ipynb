## Cleaning Our Data - Part of EDA

- In this step we perform the following:
    - Checking for missing values.
    - Dropping Columns we do not need
    - Checking for Duplicates
    - Checking for outliers

### Checking for missing values

```python
# Checking for missing values in the 
# summer_olympics Dataframe

summer_olympics.isna().sum()
```

```python
type(summer_olympics)
```

### Dropping Columns we do not need

```python
# Check the column names of the summer_olympics Dataframe

summer_olympics.columns
```

```python
# Viewing only the 'Unnamed: 0 colum' in the Dataframe

summer_olympics["Unnamed: 0"]
```

```python
summer_olympics.head() # Before dropping 'Unnamed: 0 column'
```

```python
# Drop the 'Unnamed: 0 column'

# inplace=True - permanently deletes the column inside the dataframe
# inplace=False - creates a new dataframe with the column removed

# axis=1 delete / drop column
# axis=0 delete / drop row

summer_olympics.drop("Unnamed: 0", axis=1, inplace=True)
```

```python
summer_olympics.head() # After dropping 'Unnamed: 0 column' using inplace=True
```
