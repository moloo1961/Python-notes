## Seattle 

```python
# Now plot the same graph for at least four other years 
# in your dataset and compare the results

fig, ax = plt.subplots(4,4, figsize=(20,20))

# Flatten grid into a single line of boxes
axes = ax.flatten()

# Loop to plot the 4 consecutive years
for i in range(16):
    start_row = i * 12
    end_row = (i + 1) * 12

    # slice the next 12 months
    year_data = seattle.iloc[start_row:end_row]

    # plot on the grid
    axes[i].plot(year_data["MONTH"], year_data["MLY-TAVG-NORMAL"], marker='*')
    axes[i].set_title(f"Year {i + 1}")
    axes[i].set_xlabel("MONTH")
    axes[i].set_ylabel("Avg Temp")

plt.tight_layout()
plt.show
```
