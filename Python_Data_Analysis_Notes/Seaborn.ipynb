# Seaborn

## Importing Libraries

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Bar Plots

```python
sns.get_dataset_names()
```

```python
# Return a path to the cache directory for example datasets

sns.get_data_home()
```

```python
# Load the tips data set
# The tips dataset is a built in sample data set in seaborn

tips = sns.load_dataset("tips")
tips
```

```python
# Take two variables from the tips dataset and draw bar plot

sns.barplot(x="day", y="tip", data=tips)
```

```python
# A bar plot of day vs total rainfall

sns.barplot(x="day", y="total_bill", data=tips)
```

### Clustered Plot

- This is a bar grouped into clusters

```python
# To pass the hue attribute to get more information based on that hue
# Hue - an attribute that adds a 3rd variable to a plot using color
# It is used to color code data points based on a specific variable

sns.barplot(x="day", y="tip", data=tips, hue="sex")
```

```python
# To change color, pass the palette attribute

sns.barplot(x="day", y="tip", data=tips, hue="sex", palette="spring")
```
