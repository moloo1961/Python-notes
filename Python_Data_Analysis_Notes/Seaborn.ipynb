### Categorical Bar Plot

- A grouped (nested) bar plot

```python
# Load data

penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
# Plot to show how the average body mass of penguins differ by species and sex

sns.set_theme(style="whitegrid") # sets a visual style for plots


# Draw a nested barplot by species and sex
# Catplot is a high level function used for categorical data(groups/classes)

g = sns.catplot(
    data=penguins, kind="bar", # make a bar plot
    x="species", y="body_mass_g", hue="sex", # split each species into m and f
    errorbar="sd", palette="dark", alpha=.6, height=6 
    # alpha deals with transparency 1 solid 0 invisible
    
)
g.despine(left=True) # removes top and right borders
g.set_axis_labels("", "Body mass (g)") # x axis empty y axis body mass
g.legend.set_title("") # removes legend title
```

```python
# Adjusting the color brightness

sns.barplot(x='size',y='tip',data=tips,color='magenta',capsize=0.15,saturation=0.1) # saturation- how strong the color looks
```

## Seaborn vs Matplotlib

```python
# Create Data To Plot

rng = np.random.RandomState(0)
x = np.linspace(0, 10, 500)
y = np.cumsum(rng.randn(500, 6), 0)
```

### Plot the data with Matplotlib defaults

```python
# Matplotlib

plt.plot(x, y)
plt.legend("ABCDEF", ncol=2, loc ="upper left");
```

### Plot the data with Seaborn defaults

- Seaborn gets simple matplotlib scripts to produce vastly superior output.

  - We set the stle by calling Seaborn's **set()** method.
  - By conversion, Seaborn is imported as **sns**

```python
# Rerrun the same plot above and see the difference
# Seaborn
sns.set()
plt.plot(x, y)
plt.legend('ABCDEF', ncol=2, loc= 'upper left');
```

### Exploring Seaborn Plots

- Histograms
- KDE
- Densities

- To show probability distribution, histogram focus on bin while kde(kernel density estimation) plots focus on probability density

#### Matplotlib Histogram

```python
data = np.random.multivariate_normal([0,0], [[5,2], [2,2]], size=2000)
data = pd.DataFrame(data, columns=["x", "y"])

for col in "xy":
    plt.hist(data[col], density=True, alpha=0.5)
```

### Seaborn kdeplot

```python
for col in 'xy':
    sns.kdeplot(data[col], fill= True)
```
