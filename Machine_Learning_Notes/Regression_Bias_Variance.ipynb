### Section B : Auto CV Dataset

```python
auto = pd.read_csv(r"C:\Users\User\Desktop\Data-Science-Class\Data\auto.csv")
```

```python
# Drop the "origin" column

auto = auto.drop("origin", axis=1)
```

```python
auto.head()
```

### Create the Feature and Target Variable

```python
X = auto.drop("mpg", axis=1)
y = auto["mpg"]
```

```python
X.head()
```

```python
y.head()
```

### Making a Prediction with a Single Feature

```python
X_hp = X[["hp"]]
X_hp.head()
```

```python
plt.scatter(X_hp, y)
plt.ylabel("Miles Per Gallon (mpg)")
plt.xlabel("Horsepower (hp)")
plt.show()
```

```python
reg = LinearRegression()
reg.fit(X_hp, y)
predictions = reg.predict(X_hp)

plt.scatter(X_hp, y)
plt.plot(X_hp, predictions, color="red")
plt.ylabel("Miles Per Gallon (mpg)")
plt.xlabel("Horsepower (hp)")
plt.show()
```

```python
Xhp_train, Xhp_test, y_train, y_test =train_test_split(X_hp, y, test_size=0.2, random_state=42)

reg2 = LinearRegression()
reg2.fit(Xhp_train, y_train)

predictions =reg2.predict(Xhp_test)

plt.scatter(Xhp_test, y_test)
plt.plot(Xhp_test, predictions, color="red")
plt.ylabel("Miles Per Gallon (mpg)")
plt.xlabel("Horsepower (hp)")
plt.show()
```
