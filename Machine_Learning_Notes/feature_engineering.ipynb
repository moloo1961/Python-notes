## Feature Engineering

Feature engineering is the process of creating, selecting, modifying, or transforming raw data into features that better represent the problem to predictive models, improving model performance.

---

### Why is Feature Engineering Important?

- Enhances model accuracy  
- Helps models learn patterns more easily  
- Reduces noise and irrelevant information  

---

### Steps in Feature Engineering

- Understanding the data (EDA)  
- Cleaning data (handling missing values, duplicates)  
- Creating new features (e.g., `total_price = price × quantity`)  
- Transforming features (scaling, encoding)  
- Selecting important features (feature selection)  

---

## Encoding

Categorical values represent data that can be classified into distinct categories.

### Types of Categorical Variables

- **Binary** – variables with 2 categories (e.g., yes/no, true/false)  
- **Nominal** – variables with two or more unordered categories (e.g., colors, countries)  
- **Ordinal** – variables with two or more ordered categories (e.g., education level, customer satisfaction)  
- **High Cardinality** – variables with a large number of unique categories  

---

### Encoding Techniques

#### One-Hot Encoding  
- Creates a new binary feature for each category  
- The value is 1 if the data point belongs to that category, 0 otherwise  



#### Binary Encoding  
- Assigns a unique binary code (combination of 0s and 1s) to each category 



#### Ordinal Encoding  
- Assigns numerical values that represent the **order** of the categories 



#### Label Encoding  
- Assigns a unique integer to each category  
- Simple to implement but assumes an inherent order between categories  



#### Target Encoding  
- Replaces the category with a statistic (e.g., mean) of the target variable for that category  
- Can lead to data leakage if not handled properly  



#### Frequency Encoding  
- Replaces each category with its **frequency count** in the dataset  



---

### Choice of Encoding Techniques

#### 1. Number of Unique Categories (Cardinality)

**Low cardinality (few unique values):**
- One-Hot Encoding: Simple and preserves category information  
- Binary Encoding: Reduces dimensionality compared to one-hot  

**High cardinality (many unique values):**
- Target Encoding: Uses the target variable’s mean; useful but can cause data leakage  

---

#### 2. Relationship Between Categories

**Ordinal (has meaningful order):**
- Label Encoding: Assigns ordered integers; suitable if the order matters to the model  

**Nominal (no order):**
- One-Hot Encoding: Best for preserving independence between categories  
- Binary/Target Encoding: Useful for reducing dimensionality when many categories exist
