````python
# data preprocessing
import pandas as pd  
# for linear calculations
import numpy as np    
# Plotting Graphs
import seaborn as sns 
````

````python
# load the dataset
df = pd.read_csv("Encoding Data.csv")
# displaying top 10 results
df.head(10)
````

````python
# Drop 'id' column (not predictive)
df = df.drop(columns='id')
df
````

Let's examine the columns of the dataset with different types of encoding techniques. 

```Code:``` Mapping binary features present in the dataset. 

**Label Encoding:** Label encoding algorithm is quite simple and it considers an order for encoding, Hence can be used for encoding ordinal data. 

````python
# labelEncoder present in scikitlearn library
from sklearn.preprocessing import LabelEncoder  

le = LabelEncoder()
df['ord_2_label'] = le.fit_transform(df['ord_2'])

# Displaying results
print(df[['ord_2', 'ord_2_label']].head(10))
````

**One-Hot Encoding:** To overcome the Disadvantage of Label Encoding as it considers some hierarchy in the columns which can be misleading to nominal features present in the data. we can use the One-Hot Encoding strategy. 

**One-hot encoding is processed in 2 steps:**
1. Splitting of categories into different columns.
2. Put '0 for others and '1' as an indicator for the appropriate column.

````python
df = df.drop(['ord_2'], axis=1)
df
````

One-Hot encoding with pandas 

````python
# OneHotEncoding using pandas
df = pd.get_dummies(df, columns=['nom_0'], dtype='int')

# Displaying result
print(df.head(10))
````

Ordinal Encoding

````python
# bin_1 and bin_2
df['bin_1_encoded'] = df['bin_1'].map({'F': 0, 'T': 1})
df['bin_2_encoded'] = df['bin_2'].map({'N': 0, 'Y': 1})
````

````python
df
````

````python
df = df.drop(['bin_1', 'bin_2'], axis=1)
df
````

Ordinal Encoding (custom order: Cold < Warm < Hot)

````python
# Define a custom order
ord_map = {'Cold': 0, 'Warm': 1, 'Hot': 2}
df['ord_2_ordinal'] = df['ord_2'].map(ord_map)

print(df[['ord_2', 'ord_2_ordinal']].head(10))
````

**Frequency Encoding:** We can also encode considering the frequency distribution. This method can be effective at times for nominal features.

````python
# normal count of each category
df['ord_2'].value_counts()
````

````python
# Frequency Counting (with normalize=True)
df['ord_2'].value_counts(normalize=True)
````

````python
freq_map = df['ord_2'].value_counts(normalize=True).to_dict()
df['ord_2_freq'] = df['ord_2'].map(freq_map)

print(df[['ord_2', 'ord_2_freq']].head(10))
````

Target Encoding (requires target variable)

````python
# Add a dummy target column
df['target'] = np.random.randint(0, 2, size=len(df))

# Target mean encoding
target_mean = df.groupby('ord_2')['target'].mean()
df['ord_2_target'] = df['ord_2'].map(target_mean)

print(df[['ord_2', 'target', 'ord_2_target']].head(10))
````
